```python
from sqlalchemy import create_engine
import psycopg2
import pandas as pd
import json
import os
from  pathlib import Path
from dotenv import load_dotenv
load_dotenv()

#Cria a URL de conexão  no .env(dialect+driver://user:password@host:port/dbname)
DATABASE_URL = os.getenv("DATABASE_URL")

# Cria o engine do SQLAlchemy
engine = create_engine(DATABASE_URL)
  
# Teste de conexão:
try:
    with engine.connect() as connection:
        print("Conexão estabelecida com sucesso!")
except Exception as e:
    print(f"Erro ao conectar: {e}")

query = """
SELECT 
   *
FROM public.anac
"""

df=pd.read_sql_query(query,engine)
df.head()
```

```python
#JSON:              
pasta_destino_json = r"/home/lluan/Data_Enginner/Pipeline-de-Dados-ANAC/Anac/Daframes/Anac.json" #Caminho do arquivo para o dateFrame ser salvo 
df.to_json(pasta_destino_json, orient="index")

#orient  -> Orientação do formato
#lines=True = separados por quebras da linha, só aceita no orint = records
#Tipos de de orientação = records, index, columns, value, table
```

```python
#EXCEL:
pasta_destino_excel = r"/home/lluan/Data_Enginner/Pipeline-de-Dados-ANAC/Anac/Daframes/Anac.xlsx" #Caminho do arquivo para o dateFrame salvo
df.to_excel(pasta_destino_excel,index=False, sheet_name="Base")

#sheet_name = guia da planilha
#index = indice do python
```

```python
#CSV:
pasta_destino_csv = r"/home/lluan/Data_Enginner/Pipeline-de-Dados-ANAC/Anac/Daframes/Anac.csv" #Caminho do arquivo para o dateFrame ser salvo 
df.to_csv(pasta_destino_csv,sep=',')
#sep = separador(delimitador)
```

```python
#Parquet:
pasta_destino_parquet= r"/home/lluan/Data_Enginner/Pipeline-de-Dados-ANAC/Anac/Daframes/Anac.parquet"
df.to_parquet(pasta_destino_parquet,index=False) #Salve o dataframe  em um arquivo parquet #Caminho do arquivo para o dateFrame ser salvo 

#Recomendações:
#pip install pyarrow
# pip install fastparquet
```
