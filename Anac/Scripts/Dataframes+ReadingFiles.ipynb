```python
from sqlalchemy import create_engine
import psycopg2
import pandas as pd
import json
import os
from dotenv import load_dotenv
load_dotenv()

DB_USER = os.getenv("DB_USER")
DB_PASSWORD = os.getenv("DB_PASSWORD")
DB_HOST = os.getenv("DB_HOST")
DB_PORT = os.getenv("DB_PORT")
DB_NAME = os.getenv("DB_NAME")

# Cria a URL de conexão (dialect+driver://user:password@host:port/dbname)
DATABASE_URL = f"postgresql+psycopg2://{DB_USER}:{DB_PASSWORD}@{DB_HOST}:{DB_PORT}/{DB_NAME}"

# Cria o engine do SQLAlchemy
engine = create_engine(DATABASE_URL)

# Teste de conexão:
try:
    with engine.connect() as connection:
        print("Conexão estabelecida com sucesso!")
except Exception as e:
    print(f"Erro ao conectar: {e}")

query = """
SELECT 
   "ORG_NOME",
    "PAPP_COD",
	"PAPP_NOME",
	"PAPP_PN",
	"APAA_CODI",
	"APAA_DATA",
	"APAA_STATUS"
FROM public.anac_mapeamento
"""

df=pd.read_sql_query(query,engine)
df.head()
```

```python
#JSON:              
pasta_destino_json = r"C:\Users\manue\OneDrive\Desktop\Data-Enginner\Pipeline-de-Dados-ANAC\Anac\Arquivos_Dataframes_Saves\Anac.json"
df.to_json(pasta_destino_json, orient="index")
#orient  -> Orientação do formato
#lines=True = separados por quebras da linha, só aceita no orint = records
#Tipos de de orientação = records, index, columns, value, table
```

```python
#Excel:
pasta_destino_excel = r"C:\Users\manue\OneDrive\Desktop\Data-Enginner\Pipeline-de-Dados-ANAC\Anac\Arquivos_Dataframes_Saves\Anac.xlsx"
df.to_excel(pasta_destino_excel, index = False,sheet_name='BASE')
#Sheet_name = guia da planilha
#index = indice do python
```
