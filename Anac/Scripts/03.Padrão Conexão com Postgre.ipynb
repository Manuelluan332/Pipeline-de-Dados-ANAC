```python
import   psycopg2

#Parametro de conexão:
dbname = 'seu banco'  
user = 'seu usuario'
password = 'sua senha'      
host = 'localhost'
port = '5432'

#Cria uma conexão:
conn= psycopg2.connect(
    dbname=dbname,
    user=user,
    password=password,
    host=host,
    port=port
)

#Cria uma cursor  para manipular os dados:
cur= conn.cursor()

#Verifica as alerações feitas e subir tudo para o banco de dados:
conn.commit()

#Fecha o cursor e a conexão:
cur.close()
conn.close()
```
